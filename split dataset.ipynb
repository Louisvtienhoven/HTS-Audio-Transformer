```python
import os
import shutil
import pandas as pd

# Paths
dataset_root = r"C:\Users\Louis\PycharmProjects\HTS-Audio-Transformer\anomaly_detection_workspace\Data\raw\Master"
full_dataset_path = os.path.join(dataset_root, "Dataset_final_15_second")
meta_csv_path = os.path.join(dataset_root, "meta", "meta_final_dataset_train_ver.csv")

train_output_path = os.path.join(dataset_root, "Dataset_final_train_ver")
val_output_path = os.path.join(dataset_root, "Dataset_final_val")

# Create output directories if they don't exist
os.makedirs(train_output_path, exist_ok=True)
os.makedirs(val_output_path, exist_ok=True)

# Load CSV with header
meta_df = pd.read_csv(meta_csv_path)
train_filenames = set(meta_df["filename"].tolist())

# Process each file
all_files = [f for f in os.listdir(full_dataset_path) if f.lower().endswith(".wav")]

for f in all_files:
    src = os.path.join(full_dataset_path, f)
    if f in train_filenames:
        dst = os.path.join(train_output_path, f)
    else:
        dst = os.path.join(val_output_path, f)
    shutil.copyfile(src, dst)

print(f"[INFO] Done.")
print(f" → {len(os.listdir(train_output_path))} files in Dataset_final_train_ver")
print(f" → {len(os.listdir(val_output_path))} files in Dataset_final_val")
```
