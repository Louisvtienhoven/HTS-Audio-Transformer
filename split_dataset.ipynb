```python
import os
import random
import shutil

# Set paths
base_folder = r'C:\Users\Louis\PycharmProjects\HTS-Audio-Transformer\anomaly_detection_workspace\Data\raw\Master'
input_folder = os.path.join(base_folder, 'Dataset_final_15_second')
train_ver_folder = os.path.join(base_folder, 'Dataset_final_train_ver')
val_folder = os.path.join(base_folder, 'Dataset_final_val')

# Create output folders if they don't exist
os.makedirs(train_ver_folder, exist_ok=True)
os.makedirs(val_folder, exist_ok=True)

# List all .wav files
wav_files = [f for f in os.listdir(input_folder) if f.lower().endswith('.wav')]
random.shuffle(wav_files)

# Split files
n = len(wav_files)
split_index = n // 6
val_files = wav_files[:split_index]
train_ver_files = wav_files[split_index:]

# Copy files
for f in train_ver_files:
    shutil.copy(os.path.join(input_folder, f), os.path.join(train_ver_folder, f))

for f in val_files:
    shutil.copy(os.path.join(input_folder, f), os.path.join(val_folder, f))

# Final printout
print(f"Total .wav files: {n}")
print(f"Training/Verification files: {len(train_ver_files)} -> saved to {train_ver_folder}")
print(f"Validation files: {len(val_files)} -> saved to {val_folder}")
```
