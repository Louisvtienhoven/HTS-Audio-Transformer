```python
import os
import re
import matplotlib
matplotlib.use('Qt5Agg')  # Set backend first
import matplotlib.pyplot as plt

# === Set global matplotlib styles ===
plt.rcParams.update({
    'font.size': 18,
    'axes.titlesize': 20,
    'axes.labelsize': 18,
    'legend.fontsize': 14,
    'xtick.labelsize': 16,
    'ytick.labelsize': 16,
    'lines.linewidth': 2.0,
    'legend.frameon': True,
    'font.weight': 'bold',
    'axes.labelweight': 'bold',
    'axes.titleweight': 'bold',
    'xtick.major.width': 1.5,
    'ytick.major.width': 1.5,
})

def parse_log_file(filepath):
    with open(filepath, 'r', encoding='utf-8') as f:
        lines = f.readlines()

    epoch_data = {}
    current_epoch = None
    for line in lines:
        # Match epoch number and loss
        epoch_match = re.search(r"Epoch\s+(\d+):.*?loss=([\d.eE+-]+)", line)
        if epoch_match:
            current_epoch = int(epoch_match.group(1))
            loss = float(epoch_match.group(2))

        # Match full-precision accuracy after validation
        acc_match = re.search(r"cuda:0.*?acc[\'\"]?:\s*([\d.]+)", line)
        if acc_match and current_epoch is not None:
            acc = float(acc_match.group(1))
            epoch_data[current_epoch] = (loss, acc)

    return epoch_data

def plot_metrics(log_files_by_lr):
    plt.figure(figsize=(14, 6))

    # Plot loss
    plt.subplot(1, 2, 1)
    for lr, filepath in log_files_by_lr.items():
        data = parse_log_file(filepath)
        epochs = sorted(data.keys())
        losses = [data[ep][0] for ep in epochs]
        plt.plot(epochs, losses, label=f"LR={lr}")
    #plt.title("Training Loss per Epoch", fontweight='bold')
    plt.xlabel("Epoch", fontweight='bold')
    plt.ylabel("Loss", fontweight='bold')
    plt.legend()
    plt.grid(True)

    # Plot accuracy
    plt.subplot(1, 2, 2)
    for lr, filepath in log_files_by_lr.items():
        data = parse_log_file(filepath)
        epochs = sorted(data.keys())
        accs = [data[ep][1] * 100 for ep in epochs]  # full precision
        plt.plot(epochs, accs, label=f"LR={lr}")
    plt.title("Validation Accuracy per Epoch", fontweight='bold')
    plt.xlabel("Epoch", fontweight='bold')
    plt.ylabel("Accuracy (%)", fontweight='bold')
    plt.legend()
    plt.grid(True)

    plt.tight_layout()
    plt.show()

# === Example usage ===
log_files = {
    "1e-3": "./Learning rate/log_lr_1e-3.txt",
    "1e-4": "./Learning rate/log_lr_1e-4.txt",
    # "1e-5": "./Learning rate/initial_values_high_patch",
    "original": "./Learning rate/spec384_hop40_window4"
}

plot_metrics(log_files)
```
