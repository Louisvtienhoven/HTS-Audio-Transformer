```python
import pandas as pd
import os
matplotlib.use('Qt5Agg')

# === CONFIGURATION ===
# Path to the original metadata CSV file
input_csv = r"C:\Users\Louis\PycharmProjects\HTS-Audio-Transformer\anomaly_detection_workspace\Data\raw\Master\meta\meta_final_dataset_train_ver.csv"

# === SCRIPT LOGIC ===
# Extract base name and directory
base_filename = os.path.splitext(os.path.basename(input_csv))[0]
output_dir = os.path.dirname(input_csv)

# Output file paths
batch1_csv = os.path.join(output_dir, f"{base_filename}_batch1.csv")
batch2_csv = os.path.join(output_dir, f"{base_filename}_batch2.csv")

# Load the original CSV
df = pd.read_csv(input_csv)

# Shuffle the DataFrame randomly with a fixed seed for reproducibility
df_shuffled = df.sample(frac=1.0, random_state=42).reset_index(drop=True)

# Split into two equal halves
midpoint = len(df_shuffled) // 2
df_batch1 = df_shuffled.iloc[:midpoint]
df_batch2 = df_shuffled.iloc[midpoint:]

# Save each batch to a separate CSV file
df_batch1.to_csv(batch1_csv, index=False)
df_batch2.to_csv(batch2_csv, index=False)

# Output summary
print("✅ Dataset split complete!")
print(f"🔹 Original file: {input_csv}")
print(f"🔹 Batch 1: {len(df_batch1)} samples → {batch1_csv}")
print(f"🔹 Batch 2: {len(df_batch2)} samples → {batch2_csv}")
```
